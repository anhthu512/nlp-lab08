```python
# Cell 1: Mục 12.1 - Tokenization
import json
from pathlib import Path

from pretrained_classification import LAB_DIR, MODEL_ID, MODEL_REVISION, load_tokenizer


tokenizer = load_tokenizer()
texts = ["A wonderfully acted film.", "The ending surprised me, but the plot was uneven."]
tokens = tokenizer(
    texts,
    padding=True,
    truncation=True,
    max_length=32,
    return_special_tokens_mask=True,
    return_tensors="pt",
)

keys = list(tokens.keys())
records = [
    {
        "text": text,
        "tokens": tokenizer.convert_ids_to_tokens(input_ids),
        "input_ids": input_ids,
        "attention_mask": attention_mask,
        "special_tokens_mask": special_mask,
    }
    for text, input_ids, attention_mask, special_mask in zip(
        texts,
        tokens["input_ids"].tolist(),
        tokens["attention_mask"].tolist(),
        tokens["special_tokens_mask"].tolist(),
    )
]
report = {"tokenizer": MODEL_ID, "revision": MODEL_REVISION, "keys": keys, "examples": records}
path = LAB_DIR / "outputs" / "tokenization.json"
path.parent.mkdir(parents=True, exist_ok=True)
formatted = json.dumps(report, ensure_ascii=False, indent=2)
path.write_text(formatted + "\n", encoding="utf-8")
print(formatted)
```
